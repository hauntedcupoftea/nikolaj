```python
import pandas as pd

ch = pd.read_csv("chicago.csv")
la = pd.read_csv("losangeles.csv")
ny = pd.read_csv("nycity.csv")
sf = pd.read_csv("sanfrancisco.csv")
```

```python
ch["Description"] = ch["Primary Type"] + " - " + ch["Description"]
ch = ch[["Description", "Latitude", "Longitude", "Time", "type"]]
ch["City"] = "Chicago"
ch
```

```python
# Rename Crm Cd Desc to Description
la = la.rename(columns={"Crm Cd Desc": "Description"})
la = la[["Description", "Latitude", "Longitude", "Time", "type"]]
la["City"] = "Los Angeles"
la
```

```python
ny["Description"] = ny["OFNS_DESC"] + " - " + ny["LAW_CAT_CD"]
ny = ny[["Description", "Latitude", "Longitude", "Time", "type"]]
ny["City"] = "New York"
ny
```

```python
sf["Description"] = sf["Category"] + " - " + sf["Descript"]
sf = sf[["Description", "Latitude", "Longitude", "Time", "type"]]
sf["City"] = "San Francisco"
sf
```

```python
# Are column names the same for all datasets?
print(ch.columns)
print(la.columns)
print(ny.columns)
print(sf.columns)

df = pd.concat([ch, la, ny, sf])
```

```python
df.to_csv("crime_data.csv", index=False)
```
